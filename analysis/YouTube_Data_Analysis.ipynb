```python
# Select the columns needed for YouTube analysis
youtube_df = df[[
    "video_id",
    "trending_date",
    "title",
    "channel_title",
    "category_id",
    "publish_time",
    "views",
    "likes",
    "dislikes",
    "comment_count"
]]

# Display the new DataFrame
print(youtube_df)
```

```python
# Check the number of rows and columns
print(youtube_df.shape)
```

```python
# Display the first 5 rows of the analysis DataFrame
youtube_df.head()
```

```python
# Display information about the analysis DataFrame
youtube_df.info()
```

```python
# Count how many trending videos each channel has
channel_count = youtube_df["channel_title"].value_counts()

# Display the top 10 channels
print(channel_count.head(10))
```

```python
# Convert the channel count into a DataFrame
top_channels = channel_count.head(10).reset_index()

# Rename the columns
top_channels.columns = ["channel_title", "video_count"]

# Display the result
print(top_channels)
```

```python
# Sort videos by views in descending order
top_videos = youtube_df.sort_values(by="views", ascending=False)

# Display the top 10 videos
print(top_videos[["title", "channel_title", "views"]].head(10))
```

```python
# Sort videos by likes in descending order
top_liked_videos = youtube_df.sort_values(by="likes", ascending=False)

# Display the top 10 liked videos
print(top_liked_videos[["title", "channel_title", "likes"]].head(10))
```

```python
# Sort videos by comment count in descending order
top_commented_videos = youtube_df.sort_values(by="comment_count", ascending=False)

# Display the top 10 commented videos
print(top_commented_videos[["title", "channel_title", "comment_count"]].head(10))
```

```python
# Count the number of trending videos in each category
category_count = youtube_df["category_id"].value_counts()

# Display the category counts
print(category_count)
```

```python
# Display the top 10 categories
print(category_count.head(10))
```

```python
# Convert the category count into a DataFrame
top_categories = category_count.head(10).reset_index()

# Rename the columns
top_categories.columns = ["category_id", "video_count"]

# Display the result
print(top_categories)
```

```python
# Calculate total views for each category
category_views = youtube_df.groupby("category_id")["views"].sum()

# Display the result
print(category_views)
```

```python
# Sort categories by total views
top_category_views = category_views.sort_values(ascending=False)

# Display the top 10 categories
print(top_category_views.head(10))
```

```python
# Calculate the average views for each category
category_average_views = youtube_df.groupby("category_id")["views"].mean()

# Display the top 10 categories by average views
print(category_average_views.sort_values(ascending=False).head(10))
```

```python
# Calculate total views for each channel
channel_views = youtube_df.groupby("channel_title")["views"].sum()

# Display the result
print(channel_views)
```

```python
# Sort channels by total views
top_channel_views = channel_views.sort_values(ascending=False)

# Display the top 10 channels
print(top_channel_views.head(10))
```

```python
# Calculate the average views for each channel
channel_average_views = youtube_df.groupby("channel_title")["views"].mean()

# Display the top 10 channels by average views
print(channel_average_views.sort_values(ascending=False).head(10))
```

```python
# Calculate total likes for each channel
channel_likes = youtube_df.groupby("channel_title")["likes"].sum()

# Display the top 10 channels by total likes
print(channel_likes.sort_values(ascending=False).head(10))
```

```python
# Calculate total comments for each channel
channel_comments = youtube_df.groupby("channel_title")["comment_count"].sum()

# Display the top 10 channels by total comments
print(channel_comments.sort_values(ascending=False).head(10))
```

```python
# Create a table containing channel performance
channel_performance = youtube_df.groupby("channel_title").agg({
    "views": "sum",
    "likes": "sum",
    "comment_count": "sum"
})

# Sort the table by total views
channel_performance = channel_performance.sort_values(by="views", ascending=False)

# Display the top 10 channels
print(channel_performance.head(10))
```

```python
# Count the number of trending videos for each date
trending_by_date = youtube_df["trending_date"].value_counts()

# Sort the dates
trending_by_date = trending_by_date.sort_index()

# Display the result
print(trending_by_date)
```

```python
# Display the dates with the highest number of trending videos
print(trending_by_date.sort_values(ascending=False).head(10))
```

```python
# Extract the year from publish_time
youtube_df["publish_year"] = youtube_df["publish_time"].dt.year

# Display the years
print(youtube_df["publish_year"].value_counts())
```

```python
# Extract the month from publish_time
youtube_df["publish_month"] = youtube_df["publish_time"].dt.month

# Display the number of videos published in each month
print(youtube_df["publish_month"].value_counts().sort_index())
```

```python
# Extract the day of the week from publish_time
youtube_df["publish_day"] = youtube_df["publish_time"].dt.day_name()

# Display the number of videos published on each day
print(youtube_df["publish_day"].value_counts())
```

```python
# Extract the hour from publish_time
youtube_df["publish_hour"] = youtube_df["publish_time"].dt.hour

# Display the number of videos published in each hour
print(youtube_df["publish_hour"].value_counts().sort_index())
```

```python
# Create a separate copy of the analysis DataFrame
youtube_df = youtube_df.copy()
```

```python
# Calculate total engagement using likes and comments
youtube_df["engagement"] = youtube_df["likes"] + youtube_df["comment_count"]

# Display the first 5 engagement values
print(youtube_df["engagement"].head())
```

```python
# Calculate engagement rate based on views
youtube_df["engagement_rate"] = (youtube_df["engagement"] / youtube_df["views"]) * 100

# Display the first 5 engagement rates
print(youtube_df["engagement_rate"].head())
```

```python
# Sort videos by engagement in descending order
top_engagement_videos = youtube_df.sort_values(by="engagement", ascending=False)

# Display the top 10 videos by engagement
print(top_engagement_videos[["title","channel_title","engagement"]].head(10))
```

```python
# Sort videos by engagement rate in descending order
top_engagement_rate = youtube_df.sort_values(by="engagement_rate", ascending=False)

# Display the top 10 videos by engagement rate
print(top_engagement_rate[["title","channel_title","views","engagement_rate"]].head(10))
```

```python
# Calculate the average engagement rate
average_engagement_rate = youtube_df["engagement_rate"].mean()

# Display the average engagement rate
print(average_engagement_rate)
```

```python
# Calculate the average number of comments
average_comments = youtube_df["comment_count"].mean()

# Display the average comments
print(average_comments)
```
